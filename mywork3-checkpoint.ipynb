```python
import tensorflow as tf
import matplotlib.pyplot as plt
import cv2
import numpy as np
import os
from tqdm import tqdm
from sklearn.model_selection import train_test_split
import random
import pickle
```

```python
ParentDir = r"E:\projects\Original_NN_Task\images"
Categories = ["butterfly", "cat", "chicken","cow","dog", "elephant", "horse", "sheep", "spider","squirrel"]
plt.figure(figsize = (20, 20))
y = 0
for category in Categories:
    path = os.path.join(ParentDir, category)
    """
    os.path.join() just combines folder/file names into a valid path depending on the OS. For example,
    - linux ParentDir/category
    - windows ParentDir\category
    so your project will work on any OS without problems
    """
    x = 0
    for img in os.listdir(path):
        img_array = cv2.imread(os.path.join(path, img), cv2.IMREAD_GRAYSCALE)
        if img_array is None:
            print(f"failed to load the {img}")
            continue
        plt.subplot(10, 10, y + 1)
        plt.imshow(img_array, cmap = 'gray')
        plt.axis('off')
        x+=1
        y+=1
        if x == 10:
            break
plt.tight_layout()
plt.show()
```

```python
print(img_array)
print(img_array.shape)
```

```python
IMG_SIZE = 50
new_array = cv2.resize(img_array, (IMG_SIZE, IMG_SIZE))
plt.imshow(new_array, cmap = 'gray')
plt.show()
```

```python
training_data = []
def categories_training_data():
    for category in Categories:
        path = os.path.join(ParentDir, category)
        class_num = Categories.index(category)
        for img in tqdm(os.listdir(path)):
            img_array = cv2.imread(os.path.join(path, img), cv2.IMREAD_GRAYSCALE)
            if img_array is None:
                print(f"Failed to load the image {img}")
                continue
            img_array = cv2.resize(img_array, (IMG_SIZE, IMG_SIZE))
            training_data.append([img_array, class_num])
```

```python
categories_training_data()
print(len(training_data))
```

```python
for sample1 in training_data[:10]:
    print(sample1[1])
print("--------------------------")
random.shuffle(training_data)
"""
we need to reorder the data to be in random way so that is necessary because the model learn in batchs, so it's very bad if it find batch whose all
images are dogs then another whose image are all cats
NOTE: it's already enough the randomness provided by train_test_split, but double-shuffle is the best practice
"""
for sample2 in training_data[:10]:
    print(sample2[1])
```

```python
plt.imshow(training_data[0][0], cmap = 'gray')
plt.show()
```

```python
X = []
Y = []
for features, label in training_data:
    X.append(features)
    Y.append(label)
X = np.array(X)
Y = np.array(Y)
print(X.shape)
print(Y.shape)
```

```python
X = X/255.0
```

```python
pickle_out = open(r"E:\projects\NN_Task_Files\X.pickle", "wb")
pickle.dump(X, pickle_out)
"""
pickle (.dump/.load) is more suitable to store/read content and metadata of low-level things like arrays, variables...etc., unlike standard approach
that deals with high-level and human content
"""
pickle_out.close()
pickle_out = open(r"E:\projects\NN_Task_Files\Y.pickle", "wb")
pickle.dump(Y, pickle_out)
pickle_out.close()

print("the data is successfully stored")
```
