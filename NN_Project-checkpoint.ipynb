## <a id="build"></a> Data Manipulation

```python
emotions_class = ["angry", "disgust", "fear", "happy", "neutral", "sad", "surprise"]
training_dataset = r"E:\projects\Original_NN_Project\images\train"
testing_dataset = r"E:\projects\Original_NN_Project\images\test"
IMG_SIZE = 50
```

### Create the training and testing data and show a sample of each one

```python
def load_data(IMG_SIZE, emotions_class, parentDir):
    dataset = []
    for i, folder in enumerate(emotions_class):
        plt.figure(figsize = (10, 2))
        plt.suptitle(folder)
        j = 0
        path = os.path.join(parentDir, folder)
        for img in os.listdir(path):
            img_array = os.path.join(path, img)
            img_array = cv2.imread(img_array, cv2.IMREAD_GRAYSCALE)
            if img_array is None:
                print(f"Failed to load the image: {img}")
                continue
            if j < 10:
                plt.subplot(1, 10, j + 1)
                plt.imshow(img_array, cmap = 'gray')
                plt.axis('off')
                j+=1
            img_array = cv2.resize(img_array,(IMG_SIZE, IMG_SIZE))
            img_array = img_array / 255.0
            dataset.append([img_array, i])
        plt.tight_layout()
        plt.show()
    print("the dataset successfully loaded");
    return dataset
```

```python
training_data = load_data(IMG_SIZE, emotions_class, training_dataset)
```

```python
testing_data = load_data(IMG_SIZE, emotions_class, testing_dataset)
```

### Splitting features and labels

```python
def splitting_features_label(dataset):
    X = []
    Y = []
    for (features, label) in dataset:
        X.append(features)
        Y.append(label)
    return np.array(X), np.array(Y)
```

```python
X_train, Y_train = splitting_features_label(training_data)
X_test, Y_test = splitting_features_label(testing_data)
print(X_train.shape)
print(Y_train.shape)
print(X_test.shape)
print(Y_test.shape)
```

### Store the features and labels of training and testing data in files

```python
def storing(X, Y, dataset, data_role):
    outpickle = open(rf"E:\projects\NN_Files\X_{data_role}.pickle", "wb")
    pickle.dump(X, outpickle)
    outpickle.close()
    outpickle = open(rf"E:\projects\NN_Files\Y_{data_role}.pickle", "wb")
    pickle.dump(Y, outpickle)
    outpickle.close()
    outpickle = open(rf"E:\projects\NN_Files\Dataset_{data_role}.pickle", "wb")
    pickle.dump(dataset, outpickle)
    outpickle.close()
    print("the files successfully stored")
```

```python
storing(X_train, Y_train, training_data,"training")
```

```python
storing(X_test, Y_test, testing_data,"testing")
```

### Load the files of features and labels

```python
def loading(path_X_file, path_Y_file, path_dataset_file):
    inpickle = open(path_X_file, "rb")
    X = pickle.load(inpickle)
    inpickle.close()
    inpickle = open(path_Y_file, "rb")
    Y = pickle.load(inpickle)
    inpickle.close()
    inpickle = open(path_dataset_file, "rb")
    dataset = pickle.load(inpickle)
    inpickle.close()
    print("the files successfully loading")
    return np.array(X), np.array(Y), dataset
```

## <a id="Build"></a> 🏗️ Build Model

This **CNN** model is designed for **Facial Emotion Classification** (7 classes). It features a streamlined architecture optimized for feature extraction and stability.

### **Architecture Highlights**
*   **Feature Extraction:** 3 Conv2D blocks (64, 128, 256 filters) to capture facial patterns.
*   **Normalization:** `BatchNormalization` after each convolution for faster and more stable training.
*   **Efficiency:** `MaxPooling2D` layers to reduce spatial dimensions and focus on key features.
*   **Regularization:** Uses **Dropout** (0.25 & 0.5) to prevent overfitting and improve generalization.
*   **Output:** A 256-unit Dense layer leading to a **Softmax** output for emotion probability.

### **Model Specifications**

| Feature | Details |
| :--- | :--- |
| **Model Type** | Sequential CNN |
| **Input Shape** | Gray images `(IMG_SIZE, IMG_SIZE, 1)` |
| **Classes** | 7 Emotions (Happy, Sad, Angry, etc.) |
| **Regularization** | Dropout + Batch Normalization |

---

Model 1

```python
model_1 = tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(32, (3, 3), activation='relu', input_shape=(IMG_SIZE, IMG_SIZE, 1)),
    tf.keras.layers.MaxPooling2D((2, 2)),

    tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D((2, 2)),

    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.3), 
    tf.keras.layers.Dense(7, activation='softmax')
])

model_1.summary()
```

Model 2

```python

model = tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(64, (3, 3), activation='relu', input_shape=(IMG_SIZE, IMG_SIZE, 1)),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D((2, 2)),

    tf.keras.layers.Conv2D(128, (3, 3), activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Dropout(0.25),

    tf.keras.layers.Conv2D(256, (3, 3), activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D((2, 2)),

    tf.keras.layers.Flatten(),

    tf.keras.layers.Dense(256, activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Dropout(0.5),

    tf.keras.layers.Dense(7, activation='softmax')
])

model.summary()
```

## <a id="Compile"></a> Compile

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

## <a id="Fit"></a> Fit

```python


X_train_aug = np.expand_dims(X_train, axis=-1)
X_test_aug  = np.expand_dims(X_test, axis=-1)

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',        # Watch the validation loss
    patience=3,                # If it doesn't improve for 3 epochs in a row, stop!
    restore_best_weights=True  # Automatically rewind the brain to its best epoch
)

datagen = tf.keras.preprocessing.image.ImageDataGenerator(
    rotation_range=10,         # Randomly rotate images up to 10 degrees
    zoom_range=0.1,            # Randomly zoom in up to 10%
    width_shift_range=0.1,     # Randomly shift horizontally
    height_shift_range=0.1,    # Randomly shift vertically
    horizontal_flip=True       # Randomly flip faces (left/right)
)
datagen.fit(X_train_aug)


history = model.fit(
    datagen.flow(X_train_aug, Y_train, batch_size=32),
    epochs=30,  
    validation_data=(X_test_aug, Y_test),
    callbacks=[early_stopping]
)
```
